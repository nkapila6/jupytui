```python
import pandas as pd

df = pd.DataFrame({
    "region": ["north", "south", "east", "west"],
    "sales": [120.5, 98.25, 143.0, 87.75],
    "units": [12, 9, 15, 8],
})
df
```

```python
def growth(x: float, rate: float = 0.1) -> float:
    return x * (1 + rate)

df["next_year"] = df["sales"].map(growth)
df.sort_values("sales")
```

```python
import matplotlib.pyplot as plt

df.plot.bar(x="region", y=["sales", "next_year"])
plt.title("sales by region")
plt.show()
```
